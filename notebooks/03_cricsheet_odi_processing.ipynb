```python
# Create outcome classification from original JSON records

outcome_records = []

with zipfile.ZipFile(zip_path, "r") as zip_file:

    for match_id in cricsheet_matches_df["match_id"]:

        filename = f"{match_id}.json"

        with zip_file.open(filename) as file:
            match = json.load(file)

        outcome = match["info"].get("outcome", {})

        if "winner" in outcome:
            result_type = "Win"

        elif outcome.get("result") == "tie":
            result_type = "Tie"

        elif outcome.get("result") == "no result":
            result_type = "No result"

        else:
            result_type = "Other"

        outcome_records.append({
            "match_id": match_id,
            "result_type": result_type
        })

outcome_df = pd.DataFrame(outcome_records)

cricsheet_matches_df = cricsheet_matches_df.merge(
    outcome_df,
    on="match_id",
    how="left",
    validate="one_to_one"
)

print("Outcome classification:")
print(cricsheet_matches_df["result_type"].value_counts())

print("\nTotal records:", len(cricsheet_matches_df))
```

```python
innings_records = []

with zipfile.ZipFile(zip_path, "r") as zip_file:

    for match_id in cricsheet_matches_df["match_id"]:

        filename = f"{match_id}.json"

        with zip_file.open(filename) as file:
            match = json.load(file)

        for innings_number, innings in enumerate(
            match["innings"], start=1
        ):

            total_runs = 0
            total_wickets = 0
            total_deliveries = 0

            for over in innings["overs"]:
                for delivery in over["deliveries"]:

                    total_runs += delivery["runs"]["total"]
                    total_deliveries += 1

                    total_wickets += len(
                        delivery.get("wickets", [])
                    )

            innings_records.append({
                "match_id": match_id,
                "innings_number": innings_number,
                "batting_team": innings["team"],
                "total_runs": total_runs,
                "wickets_lost": total_wickets,
                "delivery_records": total_deliveries,
                "target_runs": innings.get(
                    "target", {}
                ).get("runs")
            })

cricsheet_innings_df = pd.DataFrame(innings_records)

print("Innings dataset shape:", cricsheet_innings_df.shape)

display(cricsheet_innings_df.head(10))
```

```python
# Recalculate innings wickets correctly

innings_records = []

with zipfile.ZipFile(zip_path, "r") as zip_file:

    for match_id in cricsheet_matches_df["match_id"]:

        filename = f"{match_id}.json"

        with zip_file.open(filename) as file:
            match = json.load(file)

        for innings_number, innings in enumerate(
            match["innings"], start=1
        ):

            total_runs = 0
            total_wickets = 0
            total_deliveries = 0

            for over in innings["overs"]:
                for delivery in over["deliveries"]:

                    total_runs += delivery["runs"]["total"]
                    total_deliveries += 1

                    for wicket in delivery.get("wickets", []):

                        if wicket["kind"] != "retired hurt":
                            total_wickets += 1

            innings_records.append({
                "match_id": match_id,
                "innings_number": innings_number,
                "batting_team": innings["team"],
                "total_runs": total_runs,
                "wickets_lost": total_wickets,
                "delivery_records": total_deliveries,
                "target_runs": innings.get(
                    "target", {}
                ).get("runs")
            })

cricsheet_innings_df = pd.DataFrame(innings_records)

print("Updated innings dataset:", cricsheet_innings_df.shape)

print("\nWickets lost distribution:")
print(
    cricsheet_innings_df["wickets_lost"]
    .value_counts()
    .sort_index()
)

print("\nInnings with more than 10 wickets:")
display(
    cricsheet_innings_df[
        cricsheet_innings_df["wickets_lost"] > 10
    ]
)
```

```python
dismissal_types = {}

with zipfile.ZipFile(zip_path, "r") as zip_file:

    for match_id in cricsheet_matches_df["match_id"]:

        filename = f"{match_id}.json"

        with zip_file.open(filename) as file:
            match = json.load(file)

        for innings in match["innings"]:
            for over in innings["overs"]:
                for delivery in over["deliveries"]:

                    for wicket in delivery.get("wickets", []):

                        kind = wicket["kind"]

                        dismissal_types[kind] = (
                            dismissal_types.get(kind, 0) + 1
                        )

dismissal_df = pd.DataFrame(
    list(dismissal_types.items()),
    columns=["dismissal_type", "count"]
).sort_values("count", ascending=False)

display(dismissal_df)
```

```python
import json
from collections import Counter

match_ids = set(cricsheet_matches_df["match_id"].astype(str))

source_duplicate_groups = []
source_duplicate_records = 0

with zipfile.ZipFile(zip_path, "r") as zip_ref:

    for file_name in zip_ref.namelist():

        if not file_name.endswith(".json"):
            continue

        match_id = Path(file_name).stem

        if match_id not in match_ids:
            continue

        with zip_ref.open(file_name) as file:
            match_data = json.load(file)

        for innings_number, innings in enumerate(
            match_data["innings"], start=1
        ):
            for over_data in innings["overs"]:

                deliveries = over_data["deliveries"]

                signatures = [
                    json.dumps(delivery, sort_keys=True)
                    for delivery in deliveries
                ]

                counts = Counter(signatures)

                for signature, count in counts.items():
                    if count > 1:
                        source_duplicate_groups.append({
                            "match_id": match_id,
                            "innings_number": innings_number,
                            "over": over_data["over"],
                            "duplicate_count": count - 1,
                            "delivery": signature
                        })

                        source_duplicate_records += count - 1

print("Duplicate groups in original JSON:", len(source_duplicate_groups))
print("Extra duplicate delivery entries in original JSON:", source_duplicate_records)

display(source_duplicate_groups[:10])
```

```python
import json
from collections import Counter, defaultdict

mismatch_groups = []

with zipfile.ZipFile(zip_path, "r") as zip_ref:

    for file_name in zip_ref.namelist():

        if not file_name.endswith(".json"):
            continue

        match_id = Path(file_name).stem

        if match_id not in match_ids:
            continue

        with zip_ref.open(file_name) as file:
            match_data = json.load(file)

        for innings_number, innings in enumerate(
            match_data["innings"], start=1
        ):

            for over_data in innings["overs"]:

                projected_groups = defaultdict(list)

                for delivery in over_data["deliveries"]:

                    runs = delivery["runs"]

                    projected = (
                        match_id,
                        innings_number,
                        innings["team"],
                        over_data["over"],
                        delivery["actual_delivery"],
                        delivery["batter"],
                        delivery["bowler"],
                        delivery["non_striker"],
                        runs["batter"],
                        runs["extras"],
                        runs["total"],
                        json.dumps(delivery.get("extras"), sort_keys=True),
                        json.dumps(delivery.get("wickets"), sort_keys=True)
                    )

                    projected_groups[projected].append(delivery)

                for projected, objects in projected_groups.items():

                    if len(objects) > 1:
                        full_signatures = {
                            json.dumps(obj, sort_keys=True)
                            for obj in objects
                        }

                        if len(full_signatures) > 1:
                            mismatch_groups.append({
                                "match_id": match_id,
                                "innings_number": innings_number,
                                "over": over_data["over"],
                                "delivery": projected[4],
                                "records": objects
                            })

print("Groups with matching extracted fields but different original JSON:",
      len(mismatch_groups))

for group in mismatch_groups[:10]:
    print("\nMatch:", group["match_id"])
    print("Innings:", group["innings_number"])
    print("Over:", group["over"])
    print("Delivery:", group["delivery"])

    for record in group["records"]:
        print(record)
```

```python
import json

# Create a copy of the original dataset
cricsheet_deliveries_clean_df = cricsheet_deliveries_df.copy()

# Convert every row into a comparable string
row_signatures = cricsheet_deliveries_clean_df.apply(
    lambda row: json.dumps(
        {
            col: row[col] if not isinstance(row[col], (dict, list))
            else row[col]
            for col in cricsheet_deliveries_clean_df.columns
        },
        sort_keys=True,
        default=str
    ),
    axis=1
)

rows_before = len(cricsheet_deliveries_clean_df)

# Keep only the first occurrence of each identical row
cricsheet_deliveries_clean_df = (
    cricsheet_deliveries_clean_df.loc[
        ~row_signatures.duplicated()
    ].reset_index(drop=True)
)

rows_after = len(cricsheet_deliveries_clean_df)

print("Rows before cleaning:", rows_before)
print("Rows after cleaning:", rows_after)
print("Duplicate rows removed:", rows_before - rows_after)
print("Remaining duplicate signatures:", row_signatures[
    ~row_signatures.duplicated()
].duplicated().sum())
```

```python
# Calculate total runs from cleaned ball-by-ball records
delivery_totals = (
    cricsheet_deliveries_clean_df
    .groupby(["match_id", "innings_number"])["total_runs"]
    .sum()
    .reset_index(name="delivery_total_runs")
)

# Compare with innings-level dataset
innings_comparison_df = cricsheet_innings_df.merge(
    delivery_totals,
    on=["match_id", "innings_number"],
    how="left"
)

# Calculate difference
innings_comparison_df["run_difference"] = (
    innings_comparison_df["total_runs"]
    - innings_comparison_df["delivery_total_runs"]
)

# Display validation results
print("Total innings checked:", len(innings_comparison_df))

print(
    "Innings with matching totals:",
    (innings_comparison_df["run_difference"] == 0).sum()
)

print(
    "Innings with mismatched totals:",
    (innings_comparison_df["run_difference"] != 0).sum()
)

display(
    innings_comparison_df[
        innings_comparison_df["run_difference"] != 0
    ].head(20)
)
```

```python
# Restore all original delivery records
cricsheet_deliveries_clean_df = cricsheet_deliveries_df.copy()

print("Restored dataset shape:", cricsheet_deliveries_clean_df.shape)

# Recalculate innings totals
delivery_totals = (
    cricsheet_deliveries_clean_df
    .groupby(["match_id", "innings_number"])["total_runs"]
    .sum()
    .reset_index(name="delivery_total_runs")
)

innings_comparison_df = cricsheet_innings_df.merge(
    delivery_totals,
    on=["match_id", "innings_number"],
    how="left"
)

innings_comparison_df["run_difference"] = (
    innings_comparison_df["total_runs"]
    - innings_comparison_df["delivery_total_runs"]
)

print(
    "Innings with matching totals:",
    (innings_comparison_df["run_difference"] == 0).sum()
)

print(
    "Innings with mismatched totals:",
    (innings_comparison_df["run_difference"] != 0).sum()
)

display(
    innings_comparison_df[
        innings_comparison_df["run_difference"] != 0
    ].head(20)
)
```

```python
# Count actual team wickets from ball-by-ball records
def count_team_wickets(wickets):
    if not isinstance(wickets, list):
        return 0

    return sum(
        1 for wicket in wickets
        if wicket.get("kind") != "retired hurt"
    )


cricsheet_deliveries_clean_df["team_wickets"] = (
    cricsheet_deliveries_clean_df["wickets"].apply(
        count_team_wickets
    )
)

# Calculate wickets per innings
delivery_wickets = (
    cricsheet_deliveries_clean_df
    .groupby(["match_id", "innings_number"])["team_wickets"]
    .sum()
    .reset_index(name="delivery_wickets")
)

# Compare with innings-level wickets
wicket_comparison_df = cricsheet_innings_df.merge(
    delivery_wickets,
    on=["match_id", "innings_number"],
    how="left"
)

wicket_comparison_df["wicket_difference"] = (
    wicket_comparison_df["wickets_lost"]
    - wicket_comparison_df["delivery_wickets"]
)

print("Total innings checked:", len(wicket_comparison_df))

print(
    "Matching wicket counts:",
    (wicket_comparison_df["wicket_difference"] == 0).sum()
)

print(
    "Mismatched wicket counts:",
    (wicket_comparison_df["wicket_difference"] != 0).sum()
)

display(
    wicket_comparison_df[
        wicket_comparison_df["wicket_difference"] != 0
    ].head(20)
)
```
