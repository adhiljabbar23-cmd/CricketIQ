```python
from urllib.parse import urljoin
import time
import pandas as pd
from bs4 import BeautifulSoup

base_url = "https://crickrida.com"
next_url = urljoin(base_url, "/matches/")

all_matches = []
visited_pages = set()

while next_url and next_url not in visited_pages:

    visited_pages.add(next_url)

    response = requests.get(
        next_url,
        headers={"User-Agent": "Mozilla/5.0"},
        timeout=30
    )

    print("Reading:", next_url, "| Status:", response.status_code)

    if response.status_code != 200:
        break

    soup = BeautifulSoup(response.text, "html.parser")
    table = soup.find("table", class_="score-table")

    if table is None:
        print("Match table not found. Stopping.")
        break

    for row in table.find("tbody").find_all("tr"):

        date_cell = row.find("th")
        data_cells = row.find_all("td")

        if date_cell is None or len(data_cells) < 4:
            continue

        match_link = data_cells[0].find("a", href=True)

        all_matches.append({
            "date": date_cell.get_text(strip=True),
            "match": data_cells[0].get_text(" ", strip=True),
            "format": data_cells[1].get_text(" ", strip=True),
            "result": data_cells[2].get_text(" ", strip=True),
            "coverage": data_cells[3].get_text(" ", strip=True),
            "match_url": (
                urljoin(base_url, match_link["href"])
                if match_link else None
            )
        })

    next_link = soup.find(
        "a",
        string=lambda value: value and "Next" in value
    )

    next_url = (
        urljoin(base_url, next_link["href"])
        if next_link else None
    )

    time.sleep(1)

matches_listing_df = pd.DataFrame(all_matches)

print("\nPages collected:", len(visited_pages))
print("Total records:", len(matches_listing_df))

display(matches_listing_df.head())
```

```python
def extract_scorecard_tables(html):

    tables = pd.read_html(StringIO(html))

    batting_tables = []
    bowling_tables = []

    for table in tables:

        columns = set(table.columns)

        # Identify batting scorecards
        if {"Batter", "Dismissal", "R", "B"}.issubset(columns):
            batting_tables.append(table)

        # Identify bowling scorecards
        elif {"Bowler", "O", "M", "W", "Econ"}.issubset(columns):
            bowling_tables.append(table)

    return batting_tables, bowling_tables
```

```python
def extract_match_scorecard(match_row):

    url = match_row["match_url"]

    response = requests.get(
        url,
        headers={"User-Agent": "Mozilla/5.0"},
        timeout=30
    )

    response.raise_for_status()

    batting_tables, bowling_tables = extract_scorecard_tables(
        response.text
    )

    # Match-level information from our match listing
    match_details = {
        "date": match_row["date"],
        "match": match_row["match"],
        "format": match_row["format"],
        "result": match_row["result"],
        "match_url": url,
        "batting_innings_count": len(batting_tables),
        "bowling_innings_count": len(bowling_tables),
        "scorecard_status": "Available"
    }

    # Combine batting scorecards
    batting_records = []

    for innings_no, table in enumerate(batting_tables, start=1):

        innings_df = table.copy()
        innings_df["innings_number"] = innings_no
        innings_df["date"] = match_row["date"]
        innings_df["match_url"] = url

        batting_records.append(innings_df)

    # Combine bowling scorecards
    bowling_records = []

    for innings_no, table in enumerate(bowling_tables, start=1):

        innings_df = table.copy()
        innings_df["innings_number"] = innings_no
        innings_df["date"] = match_row["date"]
        innings_df["match_url"] = url

        bowling_records.append(innings_df)

    batting_df = (
        pd.concat(batting_records, ignore_index=True)
        if batting_records else pd.DataFrame()
    )

    bowling_df = (
        pd.concat(bowling_records, ignore_index=True)
        if bowling_records else pd.DataFrame()
    )

    return match_details, batting_df, bowling_df
```

```python
def clean_batter_name(df):

    df = df.copy()

    # Convert both columns to string safely
    batter = df["Batter"].fillna("").astype(str)
    dismissal = df["Dismissal"].fillna("").astype(str)

    # Remove dismissal text from the batter column
    df["Batter"] = [
        name.replace(dismissal_text, "", 1).strip()
        if dismissal_text else name.strip()
        for name, dismissal_text in zip(batter, dismissal)
    ]

    # Replace empty names with missing values
    df["Batter"] = df["Batter"].replace("", pd.NA)

    return df
```

```python
def extract_match_scorecard(match_row):

    url = match_row["match_url"]

    response = requests.get(
        url,
        headers={"User-Agent": "Mozilla/5.0"},
        timeout=30
    )

    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    page_text = soup.get_text(" ", strip=True)

    # Check whether the match has no recorded innings
    no_innings = (
        "This match has no recorded innings" in page_text
    )

    if no_innings:

        match_details = {
            "date": match_row["date"],
            "match": match_row["match"],
            "format": match_row["format"],
            "result": match_row["result"],
            "match_url": url,
            "batting_innings_count": 0,
            "bowling_innings_count": 0,
            "scorecard_status": "No innings recorded"
        }

        return (
            match_details,
            pd.DataFrame(),
            pd.DataFrame()
        )

    # Normal scorecard extraction
    batting_tables, bowling_tables = extract_scorecard_tables(
        response.text
    )

    match_details = {
        "date": match_row["date"],
        "match": match_row["match"],
        "format": match_row["format"],
        "result": match_row["result"],
        "match_url": url,
        "batting_innings_count": len(batting_tables),
        "bowling_innings_count": len(bowling_tables),
        "scorecard_status": (
            "Available"
            if batting_tables or bowling_tables
            else "No scorecard tables"
        )
    }

    batting_records = []

    for innings_no, table in enumerate(batting_tables, start=1):
        innings_df = table.copy()
        innings_df["innings_number"] = innings_no
        innings_df["date"] = match_row["date"]
        innings_df["match_url"] = url
        batting_records.append(innings_df)

    bowling_records = []

    for innings_no, table in enumerate(bowling_tables, start=1):
        innings_df = table.copy()
        innings_df["innings_number"] = innings_no
        innings_df["date"] = match_row["date"]
        innings_df["match_url"] = url
        bowling_records.append(innings_df)

    batting_df = (
        pd.concat(batting_records, ignore_index=True)
        if batting_records else pd.DataFrame()
    )

    bowling_df = (
        pd.concat(bowling_records, ignore_index=True)
        if bowling_records else pd.DataFrame()
    )

    return match_details, batting_df, bowling_df
```

```python
import re

def extract_innings_totals(page_text):

    pattern = (
        r"Extras\s+(\d+)\s*\((.*?)\)\s*"
        r"Total\s+(\d+)(?:/(\d+))?"
    )

    matches = re.findall(
        pattern,
        page_text,
        flags=re.IGNORECASE
    )

    results = []

    for innings_number, match in enumerate(matches, start=1):

        extras, breakdown, total, wickets = match

        extras_values = {}

        for item in breakdown.split(","):
            item = item.strip()

            extra_match = re.match(
                r"([a-z/]+)\s+(\d+)",
                item,
                flags=re.IGNORECASE
            )

            if extra_match:
                category = extra_match.group(1).lower()
                value = int(extra_match.group(2))
                extras_values[category] = value

        results.append({
            "innings_number": innings_number,
            "official_total": int(total),
            "wickets": int(wickets) if wickets else None,
            "total_extras": int(extras),
            "wides": extras_values.get("w", 0),
            "noballs": extras_values.get("nb", 0),
            "byes": extras_values.get("b", 0),
            "leg_byes": extras_values.get("lb", 0),
            "byes_legbyes_combined": extras_values.get("b/lb", 0),
            "penalty": extras_values.get("p", 0),
            "extras_breakdown": breakdown
        })

    return results
```

```python
# 1. Aggregate batting statistics
batting_innings_summary = (
    batting_clean_df
    .groupby(["match_url", "innings_number"])
    .agg(
        batter_runs=("runs", "sum"),
        balls_faced=("B", "sum"),
        fours=("4s", "sum"),
        sixes=("6s", "sum")
    )
    .reset_index()
)

# 2. Convert bowling overs into legal balls
def overs_to_balls(overs):
    overs = float(overs)
    complete_overs = int(overs)
    remaining_balls = round((overs - complete_overs) * 10)

    if remaining_balls < 0 or remaining_balls > 5:
        return pd.NA

    return complete_overs * 6 + remaining_balls

bowling_clean_df["legal_balls"] = (
    bowling_clean_df["O"].apply(overs_to_balls)
)

# 3. Aggregate bowling statistics
bowling_innings_summary = (
    bowling_clean_df
    .groupby(["match_url", "innings_number"])
    .agg(
        legal_balls_bowled=("legal_balls", "sum"),
        runs_conceded=("R", "sum"),
        wickets_taken=("W", "sum"),
        wides=("WD", "sum"),
        noballs=("NB", "sum")
    )
    .reset_index()
)

print("Batting innings summary:")
display(batting_innings_summary.head())

print("\nBowling innings summary:")
display(bowling_innings_summary.head())
```

```python
def extract_innings_teams(page_text):
    lines = [
        line.strip()
        for line in page_text.splitlines()
        if line.strip()
    ]

    innings_teams = {}

    for i, line in enumerate(lines):
        match = re.fullmatch(
            r"INNINGS\s+([12])",
            line,
            flags=re.IGNORECASE
        )

        if match and i + 1 < len(lines):
            innings_number = int(match.group(1))
            team = lines[i + 1]

            innings_teams[innings_number] = team

    return innings_teams


sample_teams = extract_innings_teams(page_text)

print(sample_teams)
```

```python
def extract_cricbuzz_match_metadata(soup):

    metadata = {
        "match_title": None,
        "date_text": None,
        "venue_full": None,
        "venue": None,
        "city": None,
        "scorecard_url": None
    }

    # Match title
    heading = soup.find("h1")

    if heading:
        metadata["match_title"] = heading.get_text(
            " ", strip=True
        )

    # Find individual information rows
    for row in soup.select("div.facts-row-grid"):

        row_text = row.get_text(" ", strip=True)

        if row_text.startswith("Date"):
            metadata["date_text"] = row_text.replace(
                "Date", "", 1
            ).strip()

        elif row_text.startswith("Venue"):
            venue_full = row_text.replace(
                "Venue", "", 1
            ).strip()

            metadata["venue_full"] = venue_full

            metadata["venue"], metadata["city"] = (
                split_venue_city(venue_full)
            )

    metadata["scorecard_url"] = (
        extract_cricbuzz_scorecard_link(soup)
    )

    return metadata
```

```python
def extract_cricbuzz_series_venues(series_url, target_series_id):

    response = requests.get(
        series_url,
        headers=headers,
        timeout=20
    )

    if response.status_code != 200:
        print("Request failed:", response.status_code)
        return pd.DataFrame()

    blocks = re.findall(
        r'\\"matchInfo\\":\{(.*?)(?=,\\"matchScore\\"|\}\}\])',
        response.text,
        flags=re.DOTALL
    )

    records = []

    for block in blocks:

        match_id = re.search(r'\\"matchId\\":(\d+)', block)
        series_id = re.search(r'\\"seriesId\\":(\d+)', block)
        match_desc = re.search(r'\\"matchDesc\\":\\"(.*?)\\"', block)
        start_date = re.search(r'\\"startDate\\":\\"(\d+)\\"', block)
        ground = re.search(r'\\"ground\\":\\"(.*?)\\"', block)
        city = re.search(r'\\"city\\":\\"(.*?)\\"', block)

        if not match_id or not series_id:
            continue

        if series_id.group(1) != str(target_series_id):
            continue

        records.append({
            "match_id": match_id.group(1),
            "series_id": series_id.group(1),
            "match": match_desc.group(1) if match_desc else None,
            "date": pd.to_datetime(
                int(start_date.group(1)), unit="ms"
            ) if start_date else pd.NaT,
            "venue": ground.group(1) if ground else None,
            "city": city.group(1) if city else None
        })

    return pd.DataFrame(records)
```

```python
def extract_crickrida_venue(page_text):
    lines = [
        line.strip()
        for line in page_text.splitlines()
        if line.strip()
    ]

    date_pattern = re.compile(
        r"\d{1,2}\s+[A-Za-z]+\s+\d{4}"
    )

    for i, line in enumerate(lines):
        if date_pattern.fullmatch(line):
            if i + 1 < len(lines):
                return lines[i + 1]

    return None


venue_test = extract_crickrida_venue(page_text)

print("Extracted venue:", venue_test)
```

```python
def recover_crickrida_venue(url):
    try:
        response = requests.get(
            url,
            headers=headers,
            timeout=20
        )

        if response.status_code != 200:
            return {
                "venue": None,
                "http_status": response.status_code,
                "recovery_status": "Request failed"
            }

        soup = BeautifulSoup(response.text, "html.parser")
        page_text = soup.get_text("\n", strip=True)

        venue = extract_crickrida_venue(page_text)

        return {
            "venue": venue,
            "http_status": response.status_code,
            "recovery_status": (
                "Recovered" if venue else "Venue not found"
            )
        }

    except requests.RequestException as error:
        return {
            "venue": None,
            "http_status": None,
            "recovery_status": f"Request error: {error}"
        }
```

```python
import re
from bs4 import BeautifulSoup
import requests
import pandas as pd

# --------------------------------------------------
# PART A: Recover result margins from result text
# --------------------------------------------------

def extract_result_margin(result):
    if pd.isna(result):
        return None, None

    result_text = str(result).strip()

    match = re.search(
        r"\bwon by\s+(\d+)\s+(runs?|wickets?)\b",
        result_text,
        flags=re.IGNORECASE
    )

    if match:
        margin = int(match.group(1))
        margin_type = match.group(2).lower()

        if margin_type.startswith("run"):
            margin_type = "runs"
        else:
            margin_type = "wickets"

        return margin, margin_type

    return None, None


margin_results = match_level_df["result"].apply(
    extract_result_margin
)

match_level_df["result_margin"] = margin_results.apply(
    lambda x: x[0]
)

match_level_df["result_margin_type"] = margin_results.apply(
    lambda x: x[1]
)

print("RESULT MARGIN RECOVERY")
print("----------------------")
print("Recovered margins:", match_level_df["result_margin"].notna().sum())

print("\nMargin types:")
display(match_level_df["result_margin_type"].value_counts(dropna=False))

print("\nWins/losses where margin could not be extracted:")
display(
    match_level_df[
        match_level_df["result_type"].isin(
            ["Afghanistan win", "Afghanistan loss"]
        )
        & match_level_df["result_margin"].isna()
    ][["date", "match", "result", "result_type"]]
)

# --------------------------------------------------
# PART B: Inspect toss information on original pages
# --------------------------------------------------

def inspect_toss_information(url):
    response = requests.get(
        url,
        headers=headers,
        timeout=20
    )

    print("\nHTTP status:", response.status_code)

    if response.status_code != 200:
        print("Page could not be retrieved.")
        return

    soup = BeautifulSoup(response.text, "html.parser")

    # Extract visible text
    lines = [
        line.strip()
        for line in soup.get_text("\n").splitlines()
        if line.strip()
    ]

    # Find toss-related lines and nearby context
    toss_positions = [
        i for i, line in enumerate(lines)
        if "toss" in line.lower()
    ]

    if not toss_positions:
        print("No visible toss information found.")
        return

    for position in toss_positions:
        start = max(0, position - 2)
        end = min(len(lines), position + 4)

        print("\nToss-related page text:")
        print("\n".join(lines[start:end]))
        print("-" * 50)


# Inspect a normal completed match and the known toss record
sample_matches = [
    "2010-07-03",
    "2017-06-14",
    "2018-09-25"
]

for sample_date in sample_matches:
    sample_row = match_level_df[
        match_level_df["date"].eq(pd.Timestamp(sample_date))
    ]

    if not sample_row.empty:
        print("\n" + "=" * 60)
        print("MATCH:", sample_row.iloc[0]["match"])
        print("DATE:", sample_date)

        inspect_toss_information(
            sample_row.iloc[0]["match_url"]
        )
```

```python
import pandas as pd
import requests
import re
from bs4 import BeautifulSoup
from IPython.display import display

# Cricbuzz URLs confirmed or previously inspected
cricbuzz_url_map = {
    "2010-07-03": (
        "https://www.cricbuzz.com/live-cricket-scorecard/9883/"
        "afg-vs-ire-4th-match-icc-world-cricket-league-division-one-2010"
    ),
    "2017-06-14": (
        "https://www.cricbuzz.com/live-cricket-scores/18293/"
        "afghanistan-tour-of-west-indies-2017-3rd-odi"
    ),
    "2018-09-25": (
        "https://www.cricbuzz.com/live-cricket-scores/20749/"
        "afg-vs-ind-super-four-match-5-asia-cup-2018"
    ),
    "2024-03-07": (
        "https://www.cricbuzz.com/cricket-match-facts/87922/"
        "afg-vs-ire-1st-odi-afghanistan-v-ireland-in-uae-2024"
    ),
    "2024-03-12": (
        "https://www.cricbuzz.com/cricket-match-facts/87936/"
        "afghanistan-vs-ireland-3rd-odi"
    ),
    "2026-10-23": (
        "https://www.cricbuzz.com/cricket-match-facts/172691/"
        "afg-vs-zim-final-odi-tri-nation-series-in-uae-2026"
    )
}


def extract_cricbuzz_toss(url):
    try:
        response = requests.get(url, headers=headers, timeout=25)
        response.raise_for_status()

        soup = BeautifulSoup(response.text, "html.parser")

        lines = [
            line.strip()
            for line in soup.get_text("\n").splitlines()
            if line.strip()
        ]

        for i, line in enumerate(lines):
            if line.lower() == "toss" and i + 1 < len(lines):
                toss_text = lines[i + 1]

                match = re.search(
                    r"(.+?) won the toss and "
                    r"(?:opt(?:ed)? to|elected to)\s+(bat|bowl|field)",
                    toss_text,
                    flags=re.IGNORECASE
                )

                if match:
                    decision = match.group(2).lower()

                    if decision == "field":
                        decision = "bowl"

                    return {
                        "toss_winner": match.group(1).strip(),
                        "toss_decision": decision,
                        "toss_status": "Extracted - pending verification",
                        "toss_source": "Cricbuzz",
                        "toss_source_url": url
                    }

        # Check whether the page explicitly says no toss took place
        page_text = " ".join(lines).lower()

        if (
            "without a ball being bowled" in page_text
            and "with toss" not in page_text
        ):
            return {
                "toss_winner": None,
                "toss_decision": None,
                "toss_status": "Possible no toss - verify",
                "toss_source": "Cricbuzz",
                "toss_source_url": url
            }

        return {
            "toss_winner": None,
            "toss_decision": None,
            "toss_status": "Not found",
            "toss_source": "Cricbuzz",
            "toss_source_url": url
        }

    except Exception as error:
        return {
            "toss_winner": None,
            "toss_decision": None,
            "toss_status": f"Request error: {str(error)[:100]}",
            "toss_source": "Cricbuzz",
            "toss_source_url": url
        }


# Extract all known URLs
bulk_toss_records = []

for date_text, url in cricbuzz_url_map.items():
    result = extract_cricbuzz_toss(url)
    result["date"] = pd.Timestamp(date_text)
    bulk_toss_records.append(result)

cricbuzz_toss_df = pd.DataFrame(bulk_toss_records)

display(
    cricbuzz_toss_df[
        ["date", "toss_winner", "toss_decision", "toss_status"]
    ].sort_values("date")
)
```

```python
# Validate toss winners
def validate_toss_winner(row):
    winner = row["toss_winner"]

    if pd.isna(winner):
        return "No toss / missing"

    if winner in [row["team_1"], row["team_2"]]:
        return "Valid"

    return "INVALID - winner not in match"


# Validate toss decisions
def validate_toss_decision(value):
    if pd.isna(value):
        return "Missing / not applicable"

    if value in ["bat", "bowl"]:
        return "Valid"

    return "INVALID"


match_level_df["toss_validation"] = match_level_df.apply(
    validate_toss_winner,
    axis=1
)

match_level_df["toss_decision_validation"] = (
    match_level_df["toss_decision"].apply(validate_toss_decision)
)


print("TOSS WINNER VALIDATION")
display(
    match_level_df["toss_validation"].value_counts(dropna=False)
)

print("\nTOSS DECISION VALIDATION")
display(
    match_level_df["toss_decision_validation"].value_counts(dropna=False)
)

print("\nINVALID RECORDS")
display(
    match_level_df[
        match_level_df["toss_validation"].str.startswith("INVALID")
        | (match_level_df["toss_decision_validation"] == "INVALID")
    ][
        ["date", "match", "toss_winner", "toss_decision"]
    ]
)
```

```python
import re

def extract_toss(api_data):

    notes = api_data.get("notes", [])

    toss_note = next(
        (
            note["text"]
            for note in notes
            if note.get("type") == "toss"
        ),
        None
    )

    if not toss_note:
        return {
            "toss_winner": None,
            "toss_decision": None,
            "toss_status": "Not available"
        }

    match = re.search(
        r"(.+?)\s*,?\s*elected to\s+(bat|bowl|field)\s+first",
        toss_note,
        flags=re.IGNORECASE
    )

    if not match:
        return {
            "toss_winner": None,
            "toss_decision": None,
            "toss_status": "Needs parsing"
        }

    winner = match.group(1).strip()
    decision = match.group(2).lower()

    if decision == "field":
        decision = "bowl"

    return {
        "toss_winner": winner,
        "toss_decision": decision,
        "toss_status": "Extracted from ESPN API"
    }
```

```python
import time
import re
import pandas as pd
import requests
from bs4 import BeautifulSoup
from pathlib import Path

search_folder = Path(
    r"C:\CricIQ\data\raw\afghanistan_odi\match_register"
)

search_folder.mkdir(parents=True, exist_ok=True)

checkpoint_path = (
    search_folder / "espn_afghanistan_search_checkpoint.csv"
)

pages_to_test = [0, 2, 4]

collected_records = []

for page in pages_to_test:

    url = (
        "https://search.cricinfo.com/ci/content/match/search.html"
        f"?search=Afghanistan;all=1;page={page}"
    )

    response = requests.get(
        url,
        headers=headers,
        timeout=30
    )

    print(f"Page {page}: HTTP {response.status_code}")

    if response.status_code != 200:
        print("Stopping collection.")
        break

    page_soup = BeautifulSoup(
        response.text,
        "html.parser"
    )

    links = page_soup.find_all(
        "a",
        class_="srchPlyrNmTxt"
    )

    for link in links:

        match_name = link.get_text(" ", strip=True)
        match_url = link.get("href", "")

        id_match = re.search(
            r"/match/(\d+)\.html",
            match_url
        )

        if not id_match:
            continue

        collected_records.append({
            "match_id": id_match.group(1),
            "match_name": match_name,
            "match_url": (
                "https://www.espncricinfo.com" + match_url
            ),
            "search_page": page
        })

    # Save after every successful page
    checkpoint_df = pd.DataFrame(collected_records)

    checkpoint_df = checkpoint_df.drop_duplicates(
        subset="match_id"
    )

    checkpoint_df.to_csv(
        checkpoint_path,
        index=False
    )

    print("Unique matches saved:", len(checkpoint_df))

    time.sleep(3)

print("\nCheckpoint:", checkpoint_path)
```

```python
import re

def extract_search_date(text):

    match = re.search(
        r",\s*((?:Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)"
        r"\s+\d{1,2}(?:-\d{1,2})?,\s+\d{4})$",
        text
    )

    return match.group(1) if match else None


espn_search_df["search_date"] = pd.to_datetime(
    espn_search_df["match_name"].apply(extract_search_date),
    errors="coerce"
)

print("Earliest date:", espn_search_df["search_date"].min())
print("Latest date:", espn_search_df["search_date"].max())

print("\nRecords by search page:")
display(espn_search_df.groupby("search_page").size())

print("\nOldest 15 records:")
display(
    espn_search_df.sort_values("search_date")[
        ["match_id", "search_date", "match_name"]
    ].head(15)
)
```

```python
def find_keys(data, keywords, path=""):
    results = []

    if isinstance(data, dict):
        for key, value in data.items():
            current_path = f"{path}.{key}" if path else key

            if any(word in key.lower() for word in keywords):
                results.append({
                    "path": current_path,
                    "value": value
                })

            results.extend(
                find_keys(value, keywords, current_path)
            )

    elif isinstance(data, list):
        for index, value in enumerate(data):
            results.extend(
                find_keys(value, keywords, f"{path}[{index}]")
            )

    return results


keywords = ["toss", "choice", "winner", "decision"]

toss_fields = find_keys(api_data, keywords)

for item in toss_fields:
    print("\nPath:", item["path"])
    print("Value:", item["value"])
```

```python
def extract_toss(api_data):

    notes = api_data.get("notes", [])

    toss_note = next(
        (
            note["text"]
            for note in notes
            if note.get("type") == "toss"
        ),
        None
    )

    if not toss_note:
        return {
            "toss_winner": None,
            "toss_decision": None,
            "toss_status": "Not available"
        }

    import re

    match = re.search(
        r"(.+?)\s*,?\s*elected to\s+(bat|bowl|field)\s+first",
        toss_note,
        flags=re.IGNORECASE
    )

    if not match:
        return {
            "toss_winner": None,
            "toss_decision": None,
            "toss_status": "Needs parsing"
        }

    winner = match.group(1).strip()
    decision = match.group(2).lower()

    if decision == "field":
        decision = "bowl"

    return {
        "toss_winner": winner,
        "toss_decision": decision,
        "toss_status": "Extracted from ESPN API"
    }
```

```python
def fetch_espn_match_data(match_id, series_id):

    url = (
        "https://site.api.espn.com/apis/site/v2/sports/cricket/"
        f"{series_id}/summary?event={match_id}"
    )

    try:
        response = requests.get(
            url,
            headers=headers,
            timeout=30
        )

        response.raise_for_status()
        data = response.json()

        header = data.get("header", {})
        competitions = header.get("competitions", [])

        if not competitions:
            return {
                "match_id": match_id,
                "status": "Competition metadata missing"
            }

        competition = competitions[0]
        match_format = (
            competition.get("class", {}).get("eventType")
        )

        result = {
            "match_id": str(match_id),
            "match": header.get("name"),
            "date": competition.get("date"),
            "format": match_format,
            "venue": (
                data.get("gameInfo", {})
                .get("venue", {})
                .get("fullName")
            )
        }

        if match_format != "ODI":
            result["status"] = "Not an ODI"
            return result

        result.update(extract_toss(data))
        result["status"] = "Success"

        return result

    except requests.RequestException as error:
        return {
            "match_id": str(match_id),
            "status": "Request failed",
            "error": str(error)
        }

    except (ValueError, KeyError, TypeError) as error:
        return {
            "match_id": str(match_id),
            "status": "Parsing failed",
            "error": str(error)
        }
```

```python
def extract_event_details(event):

    notes = event.get("notes", [])

    match_number = next(
        (
            note["text"]
            for note in notes
            if note.get("type") == "matchnumber"
        ),
        None
    )

    toss_note = next(
        (
            note["text"]
            for note in notes
            if note.get("type") == "toss"
        ),
        None
    )

    if match_number:
        if match_number.lower().startswith("odi"):
            match_format = "ODI"
        elif match_number.lower().startswith("test"):
            match_format = "Test"
        elif match_number.lower().startswith("t20i"):
            match_format = "T20I"
        else:
            match_format = "Unknown"
    else:
        match_format = "Unknown"

    return {
        "match_id": str(event.get("id")),
        "match": event.get("name"),
        "match_number": match_number,
        "format": match_format,
        "toss_note": toss_note
    }

print(extract_event_details(event))
```

```python
def extract_series_references(api_data):

    header = api_data.get("header", {})
    league = header.get("league", {})
    leagues = header.get("leagues", [])

    return {
        "match_id": header.get("id"),
        "tour_id": league.get("id"),
        "tour_name": league.get("name"),
        "tournament_id": (
            leagues[0].get("id") if leagues else None
        ),
        "tournament_name": (
            leagues[0].get("name") if leagues else None
        ),
        "squad_reference": None,
        "statistics_reference": None
    } | {
        "squad_reference": next(
            (
                link["href"]
                for link in league.get("links", [])
                if "squad" in link.get("rel", [])
            ),
            None
        ),
        "statistics_reference": next(
            (
                link["href"]
                for link in league.get("links", [])
                if "statistics" in link.get("rel", [])
            ),
            None
        )
    }


series_references = extract_series_references(test_api_data)

display(pd.DataFrame([series_references]))
```
