```python
# Create match keys for overlap detection

afghanistan_matches_df["match_key"] = (
    afghanistan_matches_df["date"].dt.strftime("%Y-%m-%d")
    + "_"
    + afghanistan_matches_df["team_1"]
    + "_"
    + afghanistan_matches_df["team_2"]
)

cricsheet_matches_df["match_key"] = (
    cricsheet_matches_df["date"].dt.strftime("%Y-%m-%d")
    + "_"
    + cricsheet_matches_df["team_1"]
    + "_"
    + cricsheet_matches_df["team_2"]
)

# Check exact match-key overlaps
exact_overlaps = set(
    afghanistan_matches_df["match_key"]
) & set(
    cricsheet_matches_df["match_key"]
)

# Create reverse-order keys for Cricsheet
cricsheet_matches_df["reverse_match_key"] = (
    cricsheet_matches_df["date"].dt.strftime("%Y-%m-%d")
    + "_"
    + cricsheet_matches_df["team_2"]
    + "_"
    + cricsheet_matches_df["team_1"]
)

reverse_overlaps = set(
    afghanistan_matches_df["match_key"]
) & set(
    cricsheet_matches_df["reverse_match_key"]
)

print("Exact team-order overlaps:", len(exact_overlaps))
print("Reverse team-order overlaps:", len(reverse_overlaps))
print("Total potential overlapping matches:",
      len(exact_overlaps | reverse_overlaps))
```

```python
# Check duplicates within each dataset

def find_duplicate_matches(df):
    match_keys = df.apply(
        lambda row: (
            row["date"],
            *sorted([row["team_1"], row["team_2"]])
        ),
        axis=1
    )

    return match_keys.duplicated(keep=False)


afg_duplicates = find_duplicate_matches(
    afghanistan_matches_df
)

cricsheet_duplicates = find_duplicate_matches(
    cricsheet_matches_df
)

print("AFGHANISTAN DATASET")
print("Duplicate records:", afg_duplicates.sum())

print("\nCRICSHEET DATASET")
print("Duplicate records:", cricsheet_duplicates.sum())

print("\nDuplicate Afghanistan records:")
display(
    afghanistan_matches_df.loc[afg_duplicates]
)

print("\nDuplicate Cricsheet records:")
display(
    cricsheet_matches_df.loc[cricsheet_duplicates]
)
```

```python
# Create standardised Afghanistan match-level dataset

afghanistan_standard_df = afghanistan_matches_df.copy()

afghanistan_standard_df["match_date"] = afghanistan_standard_df["date"]

# Standardise toss decisions
afghanistan_standard_df["toss_decision"] = (
    afghanistan_standard_df["toss_decision"]
    .replace({"bowl": "field"})
)

# Derive winner from Afghanistan's result perspective
afghanistan_standard_df["winner"] = np.select(
    [
        afghanistan_standard_df["result_type"] == "Afghanistan win",
        afghanistan_standard_df["result_type"] == "Afghanistan loss"
    ],
    [
        "Afghanistan",
        afghanistan_standard_df["opponent"]
    ],
    default=np.nan
)

# Standardise result categories
afghanistan_standard_df["result_type"] = (
    afghanistan_standard_df["result_type"]
    .replace({
        "Afghanistan win": "Win",
        "Afghanistan loss": "Win"
    })
)

afghanistan_standard_df["source"] = "Crickrida"

print("Afghanistan standardisation completed.")
display(
    afghanistan_standard_df[
        ["match_date", "team_1", "team_2", "winner",
         "result_type", "toss_winner", "toss_decision", "source"]
    ].head()
)
```

```python
# Create standardised Cricsheet match-level dataset

cricsheet_standard_df = cricsheet_matches_df.copy()

cricsheet_standard_df["match_date"] = cricsheet_standard_df["date"]

# Standardise toss decisions
cricsheet_standard_df["toss_decision"] = (
    cricsheet_standard_df["toss_decision"]
    .replace({"field": "field", "bat": "bat"})
)

# Standardise winner
cricsheet_standard_df["winner"] = (
    cricsheet_standard_df["winner"]
)

# Standardise result categories
cricsheet_standard_df["result_type"] = (
    cricsheet_standard_df["result_type"]
    .replace({
        "Win": "Win",
        "Tie": "Tie",
        "No result": "No result"
    })
)

cricsheet_standard_df["source"] = "Cricsheet"

print("Cricsheet standardisation completed.")

display(
    cricsheet_standard_df[
        ["match_date", "team_1", "team_2", "winner",
         "result_type", "toss_winner", "toss_decision", "source"]
    ].head()
)
```

```python
# Validate winner and result consistency

def validate_results(df, dataset_name):
    valid_teams = (
        df["winner"].isna()
        | df["winner"].eq(df["team_1"])
        | df["winner"].eq(df["team_2"])
    )

    valid_categories = df["result_type"].isin(
        ["Win", "Tie", "No result"]
    )

    winner_required = df["result_type"].eq("Win")
    no_winner_required = df["result_type"].isin(
        ["Tie", "No result"]
    )

    print(f"\n{dataset_name} VALIDATION")
    print("-" * 40)

    print("Invalid winner-team combinations:", (~valid_teams).sum())
    print("Invalid result categories:", (~valid_categories).sum())

    print(
        "Wins with missing winner:",
        (winner_required & df["winner"].isna()).sum()
    )

    print(
        "Ties/no-results with recorded winner:",
        (no_winner_required & df["winner"].notna()).sum()
    )


validate_results(
    afghanistan_standard_df,
    "AFGHANISTAN"
)

validate_results(
    cricsheet_standard_df,
    "CRICSHEET"
)
```

```python
def validate_toss(df, dataset_name):

    valid_toss_winner = (
        df["toss_winner"].isna()
        | df["toss_winner"].eq(df["team_1"])
        | df["toss_winner"].eq(df["team_2"])
    )

    valid_toss_decision = (
        df["toss_decision"].isna()
        | df["toss_decision"].isin(["bat", "field"])
    )

    print(f"\n{dataset_name} TOSS VALIDATION")
    print("-" * 40)

    print(
        "Invalid toss winners:",
        (~valid_toss_winner).sum()
    )

    print(
        "Invalid toss decisions:",
        (~valid_toss_decision).sum()
    )

    print(
        "Missing toss winners:",
        df["toss_winner"].isna().sum()
    )

    print(
        "Missing toss decisions:",
        df["toss_decision"].isna().sum()
    )


validate_toss(afghanistan_standard_df, "AFGHANISTAN")
validate_toss(cricsheet_standard_df, "CRICSHEET")
```

```python
# Verify all recovered venue-city mappings

recovered_records = integrated_matches_df[
    integrated_matches_df["venue"].isin(venue_city_mapping.keys())
]

mapping_check = (
    recovered_records.apply(
        lambda row: row["city"] == venue_city_mapping[row["venue"]],
        axis=1
    )
)

print("CITY MAPPING VERIFICATION")
print("=" * 45)

print("Records using mapped venues:", len(recovered_records))
print("Correct city mappings:", mapping_check.sum())
print("Incorrect city mappings:", (~mapping_check).sum())

print("\nRemaining missing cities:",
      integrated_matches_df["city"].isna().sum())

print("\nCity counts by source:")
display(
    integrated_matches_df.groupby("source")["city"].agg(
        total="count",
        missing=lambda x: x.isna().sum()
    )
)
```

```python
# Validate recovered cities while preserving existing source values

recovered_city_records = integrated_matches_df[
    integrated_matches_df["venue"].isin(venue_city_mapping.keys())
    & integrated_matches_df["source"].eq("Cricsheet")
]

correct_recovered = recovered_city_records.apply(
    lambda row: row["city"] == venue_city_mapping[row["venue"]],
    axis=1
)

print("FINAL CITY VALIDATION")
print("=" * 45)

print("Cricsheet records using mapped venues:",
      len(recovered_city_records))

print("Correct mappings:",
      correct_recovered.sum())

print("Incorrect mappings:",
      (~correct_recovered).sum())

print("Missing cities:",
      integrated_matches_df["city"].isna().sum())

print("\nCity values by source:")
display(
    integrated_matches_df.groupby("source")["city"].agg(
        total="count",
        missing=lambda x: x.isna().sum()
    )
)
```
